# EXPENSE TRACKER


```python
income=0
expense=[]
category=[]
dat=[]
```

```python
from datetime import date
```

```python
while True:
    print("----- EXPENSE TRACKER -----")
    print("1. ADD INCOME")
    print("2. ADD EXPENSES AND ITS CATEGORY ")
    print("3. VIEW EXPENSES")
    print("4. VIEW EXPENSES AVERAGE")
    print("5. VIEW INCOME")
    print("6. YOUR CURRENT BALANCE")
    print("7. EXIT")

    choice = int(input("enter your choice"))
    if choice >0 and choice <8:
        
        if choice ==1:
            inc = int(input("ENTER YOUR INCOME:"))
            if inc>0:
                 income =income+inc 
                 print("INCOME ADDED SUCCESFULLY")
                 print()
            else:
                print("INVALID INCOME AMOUNT")
           
            
        elif choice ==2:
              exp =int(input("ENTER YOUR EXPENSES"))
              if exp>0:
                  cat =input("ENTER YOUR CATEGORY")
                  expense.append(exp)
                  category.append(cat)
                  dat.append(date.today())
              else:
                  print("INVALID AMOUNT")
    
        elif choice == 3:
             print("expenses")
             if len(expense)>0:
                for e,c,d in zip(expense,category,dat):
                 print(f"   {e}    |   {c}   |   {d}   ")
                 print()
             else:
                 print("NO EXPENSES FOUND")
                 
    
        elif choice == 4:
             avg_exp =sum(expense)/len(expense)
             print(f"AVERAGE EXPENSE = {avg_exp}")
    
        elif choice == 5:
             print("income:",income)
    
        elif choice == 6:
             cur_bal = income - sum(expense)
             print("YOUR CURRENT BALANCE:", cur_bal)
    
        else:
            print("----THANK  YOU ----")
            break
    else:
        print("INVALID OPTION")
    
    
```
